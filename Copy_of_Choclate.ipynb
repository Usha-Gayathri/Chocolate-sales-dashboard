```python
import pandas as pd
import plotly.express as px
import os

# 1. Load Data - using YOUR exact file name
file_name = 'Chocolate Sales (2).csv' # Fixed typo here: 'Choclate' changed to 'Chocolate'

# Try to find it
if os.path.exists(f'/content/{file_name}'):
    path = f'/content/{file_name}'
elif os.path.exists(f'/content/sample_data/{file_name}'):
    path = f'/content/sample_data/{file_name}'
else:
    path = file_name
    print("⚠️ Make sure you uploaded the file first!")

df = pd.read_csv(path)

print("✅ File loaded successfully!")
print("Columns in your file:", df.columns.tolist())

# Clean column names
df.columns = df.columns.str.strip()

# Convert Date safely
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df = df.dropna(subset=['Date'])

# --- DASHBOARD 1: CONCLUSION - Best Product by Revenue ---
df1 = df.groupby('Product')['Amount'].sum().reset_index().sort_values('Amount', ascending=False)
fig1 = px.bar(df1, x='Product', y='Amount', title='Dashboard 1: Which product makes most money?', color='Amount', text='Amount')
fig1.update_layout(xaxis_tickangle=-45)
fig1.show()

# --- DASHBOARD 2: CONCLUSION - Sales Trend ---
df['Month'] = df['Date'].dt.to_period('M').astype(str)
df2 = df.groupby('Month')['Amount'].sum().reset_index()
fig2 = px.line(df2, x='Month', y='Amount', title='Dashboard 2: Are sales growing?', markers=True)
fig2.show()




# --- DASHBOARD 3: CONCLUSION - Volume vs Value ---
fig3 = px.scatter(df, x='Boxes Shipped', y='Amount', color='Product', size='Boxes Shipped', title='Dashboard 4: Do more boxes = more money?')
fig3.show()

# --- DASHBOARD 4: CONCLUSION - Highest Volume Product ---
df4 = df.groupby('Product')['Boxes Shipped'].sum().reset_index().sort_values('Boxes Shipped', ascending=False)
fig4 = px.bar(df4, x='Product', y='Boxes Shipped', title='Dashboard 5: Which product ships most boxes?', color='Boxes Shipped')
fig4.update_layout(xaxis_tickangle=-45)
fig4.show()


# --- DASHBOARD 5: Top 5 Products ---
df5 = df.groupby('Product')['Boxes Shipped'].sum().reset_index()
df5 = df5.sort_values('Boxes Shipped', ascending=False).head(5) # only top 5

fig5 = px.bar(df5, x='Product', y='Boxes Shipped',
              title='Dashboard 5: Top 5 Products by Volume',
              color='Boxes Shipped', text='Boxes Shipped')
fig5.update_layout(xaxis_tickangle=-45)
fig5.show()
```
